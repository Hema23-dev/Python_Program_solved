```python
# program to check the input list is ascending or descending if true - print yes else no
n=int(input())
lst=list(map(int,input().split()))
aclst=[]
for i in lst:
    aclst.append(i)
print(aclst)
for i in range(n):
  for j in range(i+1,n):
    if lst[i]>lst[j]:
      lst[i],lst[j]=lst[j],lst[i]
print(lst)
revlst=[]
for i in reversed(lst):
    revlst.append(i)
print(revlst)
if ((lst == aclst) or (revlst == aclst)):
  print("yes")
else:
  print("no")
```
