```python
for i in range(4):
  for j in range(i,4):   # this loop represent space of inverted right angle triangle
    print(' ',end=" ")
  for j in range(i):     # this loop represent right angle triangle from right
    print('*',end=" ")
  for j in range(i+1):   # this loop represent left angle triangle
    print('*',end=" ")
  print()
```

```python
##inverted hill pattern

n=7
for i in range(n):
  for j in range(i+1):   # this loop represent space of right angle triangle
    print(' ',end=" ")
  for j in range(i,n-1): # this loop represent 1st half inverted triangle , n-1 represent end * to have less column
    print('*',end=" ")
  for j in range(i,7):   # this loop represent 2nd half inverted triangle
    print('*',end=" ")
  print()


```

```python
## Diamond pattern

n=7
for i in range(n-1): ##(n-1) to reduce the row
  for j in range(i,n):
    print(" ",end=" ")
  for j in range(i):
    print("*",end=" ")
  for j in range(i+1):
    print("*",end=" ")
  print()
for i in range(n):
  for j in range(i+1):
    print(" ",end=" ")
  for j in range(i,n-1):
    print("*",end=" ")
  for j in range(i,n):
    print("*",end=" ")
  print()




```

```python
## butterfly pattern

n=4

for i in range(n-1):
  for j in range(i+1):
    print("*",end=" ")
  for j in range(i,n):
    print("-",end=" ")
  for j in range(i,n):
    print("-",end=" ")
  for j in range(i+1):
    print("*",end=" ")
  print()   # till this is the first half of the pattern
for i in range(n):
  for j in range(i,n):
    print("*",end=" ")
  for j in range(i+1):
    print("-",end=" ")
  for j in range(i+1):
    print("-",end=" ")
  for j in range(i,n):
    print("*",end=" ")
  print()


```

```python
### Prime Number Check -- program that checks if a number input by the user is a prime number using a for loop.

number = int(input("enter a number:"))
is_prime = True
if number <= 1:
    is_prime = False
else:
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            is_prime = False
            break
if is_prime:
    print(f"{number} is a prime number.")
else:
    print(f"{number} is not a prime number.")
```

```python
### Fibonacci Sequence-- program that prints the first n Fibonacci numbers, where n is input by the user.


n = int(input("Enter the number of Fibonacci numbers to print: "))
a, b = 0, 1
count = 0
while count < n:
    print(a)
    a, b = b, a + b
    count += 1
```

```python
## Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.

def transpose_matrix(matrix):
    transposed = [[matrix[j][i] for j in range(len(matrix))] for i in range(len(matrix[0]))]
    return transposed

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
transposed = transpose_matrix(matrix)
print("Original matrix:")
for row in matrix:
    print(row)
print("Transposed matrix:")
for row in transposed:
    print(row)
```

```python
## Create a default dictionary where each key has a default value of an empty list. Add some elements to the lists and print the dictionary.

from collections import defaultdict

default_dict = defaultdict(list)

default_dict['a'].append(1)
default_dict['a'].append(2)
default_dict['b'].append(3)
print(default_dict)
```

```python
## Write a function that takes a string and returns a dictionary with the count of each character in the string. Print the dictionary.

def char_count(string):
  char_count={}
  for char in string:
    if char in char_count:
       char_count[char]+=1
    else:
       char_count[char]=1
  return char_count

string="hema divya harikrishnan"
print(char_count(string))
```

```python
## convert json to dictionary
import json

try:
    with open('book_dict_json.json', 'r') as file:
        data_dictionary = json.load(file)

    print("JSON content successfully converted to a Python dictionary:")
    print(data_dictionary)
    print(f"Type of data_dictionary: {type(data_dictionary)}")

    # You can now access elements of the dictionary
    print(f"title: {data_dictionary['title']}")
    print(f"author: {data_dictionary['author']}")

except FileNotFoundError:
    print("Error: 'book_dict_json' not found. Please ensure the file exists.")
except json.JSONDecodeError:
    print("Error: Invalid JSON format in 'book_dict_json'.")
```
