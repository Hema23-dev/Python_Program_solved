TEMPERATURE CONVERSION:
------------------------
C ---> F : temp * 9/5 + 32


F ---> C : (temp - 32) * 5/9

```python
## Temperature conversion

def temperature_convertor(temp,unit):
   if unit == "C":
      return temp * 9/5 + 32    ## celsius to fahrenheit
   elif unit == "F":
      return (temp - 32) * 5/9   ## fahrenheit to celsius
   else:
      return None

temp=int(input())
unit=input()
temperature_convertor(temp,unit)
```

2. Password strength checker:

The password should be more than 8 letters and with one special character,one digit,one uppercas,one lowercase.

if it Satifys the above condition , then it is strong password


```python
def is_strong_password(password):
    """ This function checks whether the password is strong or not"""
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in "!@#$%^&*()_+" for char in password):
        return False
    return True

print(is_strong_password("Worldnew*25"))
print(is_strong_password("wOrldnew2567"))
```

6. A function to read a file and count the frequency of each word.

here i have created one txt file called sample.txt

```python
def count_word_freq(file_path):
    word_freq={}
    with open(file_path,'r') as file:          # open and read a file
        for line in file:
            words=line.split()
            for word in words:
                word=word.lower().strip('.,!?;:"\'')    # removes these .,!?;:"\' character in word
                word_freq[word]=word_freq.get(word,0)+1 #If key exists: Returns current count or key missing return 0
    return word_freq

file_path="/content/sample.txt"
print(count_word_freq(file_path))
```

filter() Function
------------------
The filter() function contructs an iterator from elements of an iterable for which a function returns true. It is used to filter out items fromm list (or any other iterable) based on a condition


```python
# we are haing a function called even

def even(num):
  if num%2==0:
    return True

num=[1,2,3,4,5]
list(filter(even,num))  #here the output filter the even condition and only print the value which is even
```

```python
## fiter with lambda function

numb=[5,1,3,7,9,0,3,55,100]
greater_than_five=list(filter(lambda x:x>5,numb))
print(greater_than_five)
```

```python
# fiter with lambda function and multiple condition

numb=[5,1,3,7,9,0,3,55,100]
even_and_greater_than_five=list(filter(lambda x:x>5 and x%2==0,numb))
print(even_and_greater_than_five)

```

```python
## filter() to check if the age is greater than 25 in list of dictionary

people=[
    {'name':'hema','age':20},
    {'name':'hema1','age':25},
    {'name':'hema2','age':30}
]

def age_greater_25(person):
  if person["age"] >= 25:
    return person['age']

list(filter(age_greater_25,people))
```

12. Function Decorator
------------------------
Define a decorator function that prints 'Executing function...' before executing  a function and 'Function executed.' after executing it. Apply this decorator to a function that takes a list of integers and returns their sum. Test the decorator function with different lists

```python
def my_decorator(func):
  def wrapper(numbers):
     print("Executing Function...")
     result=func(numbers)
     print("Function Executed")
     return result
  return wrapper

@my_decorator
def sum_lst(numbers):
    su=0
    for i in numbers:
      su=su+i
    return su

print(sum_lst([7,8,0,9]))



```

```python
def factorial(n):
  """
  calculate the factorial of a number n (i.e=>n!=n*n-1*n-2 ....*1)

  parameters:
  n(int): input number will be given

  Returns:
  int: factorial of the number n

  """
  if n == 0:
    return 1
  else:
    return n*factorial(n-1)


#calling the function
print(factorial(5))


# printing the doc string
print(factorial.__doc__)


```
