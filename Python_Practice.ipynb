## 1. Given two integers, n and m. The task is to check the relation between n and m. Print "less" if n < m,  "equal" if n == m, and 
## "greater" if n > m

## Example:
Input: n = 4, m = 8
Output: less
Explanation: 4 < 8 so print 'less'.


```python
n = int(input())
m = int(input())

# code here
if n < m:
    print("less")
elif n > m : 
    print("greater")
else:
    print("equal")
```

## Input: s = "DoctorPhenomenal"
## Output: DcoPeoea
Explanation: The characters at even indices are collected in their original order.



```python
s="DoctorPhenomenal"

for i in range(len(s)):
    if i % 2 == 0:
       print(s[i],end='')
    else:
        continue
```

## 4. Given a number n, print all numbers from 0 to n in increasing order in a single line using a while loop.


## Example
Input: n = 3
Output: 0 1 2 3
Explanation: Numbers from 0 to 3 in increasing order are 0 1 2 3.


```python
n = int(input())

# code here
i=0
while i <= n:
    print(i,end=' ')
    i=i+1
```

## 12. Given an integer array arr[], return the sum of all elements of arr.

Examples:

Input: arr[] = [1, 2, 3, 4]
Output: 10
Explanation: 1 + 2 + 3 + 4 = 10.


```python
arr=list(map(int,input().split()))
sum=0

for i in arr:
    sum=sum+i

print(sum)
```

## 16.Given an array arr[], return the floor value of the mean of its elements.

Examples:


Input: arr[] = [1, 3, 4, 2, 6, 5, 8, 7]
Output: 4
Explanation: Sum of the elements is 1 + 3 + 4 + 2 + 6 + 5 + 8 + 7 = 36, Mean = 36/8 = 4.5, floor(4.5) = 4.



```python
lst=list(map(int,input().split()))
total=0

for i in lst:
    total=total+i
    
print(round(total/len(lst)))
```
