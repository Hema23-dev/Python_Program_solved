## 1. Given two integers, n and m. The task is to check the relation between n and m. Print "less" if n < m,  "equal" if n == m, and 
## "greater" if n > m

## Example:
Input: n = 4, m = 8
Output: less
Explanation: 4 < 8 so print 'less'.


```python
n = int(input())
m = int(input())

# code here
if n < m:
    print("less")
elif n > m : 
    print("greater")
else:
    print("equal")
```

## 2.Given a number n, print the multiplication table from 1 to 10 for n in a single line, separated by spaces.

## Example:
Input: n = 9
Output: 9 18 27 36 45 54 63 72 81 90

```python
n=int(input())
for i in range(1,11):
    print(n*i)
```

## 3. Given a string s, return a string containing all characters present at even indices (0, 2, 4, ...). The characters must appear ## in the same order as in the original string.

## Input: s = "DoctorPhenomenal"
## Output: DcoPeoea
Explanation: The characters at even indices are collected in their original order.



```python
s="DoctorPhenomenal"

for i in range(len(s)):
    if i % 2 == 0:
       print(s[i],end='')
    else:
        continue
```

## 4. Given a number n, print all numbers from 0 to n in increasing order in a single line using a while loop.


## Example
Input: n = 3
Output: 0 1 2 3
Explanation: Numbers from 0 to 3 in increasing order are 0 1 2 3.


```python
n = int(input())

# code here
i=0
while i <= n:
    print(i,end=' ')
    i=i+1
```

## 6. This problem has no input. You need to write a function to be named as helloFunction(). The function should only prints Hello followed by a new line.

input: None
Output: Hello
Explanation: There is no input. The function prints Hello everytime.

```python
s=input()

def hellofunction():
    if s == "None":
       return "Hello"
    else:
       return f"Hello {s}"
    
print(hellofunction())
    
```
