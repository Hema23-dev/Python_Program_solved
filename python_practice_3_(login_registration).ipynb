## Getting valid mail address and password and storing it in csv file

```python
#Assignment0 Stage 1 and 2:
import csv
def write():
   f=open("details.csv","w",newline='')
   wo=csv.writer(f)
   wo.writerow(["user_id","pswd"])
   while True:
       #USER VALIDATION
       username = input("username:")
       i=1
       while i==1:
        if username.endswith("@gmail.com") == True or username.endswith("@GMAIL.COM") == True or username.endswith("yahoo.in") == True or username.endswith("YAHOO.IN") == True :
           verify=username.split("@")
           if verify[0] == '':
              print("please enter a valid gmail id ")
              username=input("usernameins:")
           elif verify[0].isalnum() == True or (verify[0].isupper() == True or verify[0].islower() == True or verify[1].isupper() == True or verify[1].islower() ==True):
              i=2
           else:
              print("please enter a valid gmail id ")
              username=input("usernameins:")
        else:
          print("please enter a valid gmail id ")
          username=input("usernameins:")
       #PASSWORD VALIDATION
       import re
       password = input("password:")
       x = True
       while x:
        if (len(password)<6 or len(password)>12):
           print("please enter a valid password ")
           password=input("password len:")
        elif not re.search("[a-z]",password):
           print("please enter a valid password ")
           password=input("password al:")
        elif not re.search("[0-9]",password):
           print("please enter a valid password ")
           password=input("password num:")
        elif not re.search("[A-Z]",password):
           print("please enter a valid password ")
           password=input("password cap:")
        elif not re.search("[!#$@%&'*+-/=?^_`.]",password):
          print("please enter a valid password ")
          password=input("password spec:")
        elif re.search("\s",password):
         print("please enter a valid password ")
         password=input("password spa:")
        else:
         x=False
         break
       # WRITING USERNAME AND PASSWORD IN CSV FILE
       wo.writerow([username,password])
       rec=input("Do you want to enter more records (Y/N):")
       if rec in "Nn":
          break
   f.close()

write()

```
