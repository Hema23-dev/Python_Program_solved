File Operation - Read and write Files
--------------------------------------
File Handling is a crucial part of any programming language. Python provides built-in functions and method to read from and write to files, both text and binary.

```python
### read a file

with open("example.txt",mode="r") as file:
  content=file.read()

print(content)

```

```python
## read a file line by line

with open("example.txt",mode="r") as file:
     for line in file:
        print(line.strip("\n"))          ## removes the new line
```

```python
## writing a file in already existing file(overwriting)
## this should not be done because the existing file data will be overrided

with open("example.txt",mode="w") as file:
      file.write("adding this line \n")
      file.write("adding few more line \n")


```

```python
## write a file(without overriding)

with open("example.txt",mode="a") as file:
     file.write("append operation is used to wrtie in already existing file \n")
```

```python
## write a list of line to a file

lines=['first line \n','second line \n','thrid line \n']

with open("example.txt",mode="a") as file:
  file.writelines(lines)
```

```python
## binary file

## writing binary in binary file


data= b"\x00\x01\x02\x03\x04"
with open("binaryfile.bin",mode="wb") as file:           ##  wb - write binary data
     file.write(data)

```

```python
## read binary file

with open("binaryfile.bin",mode="rb") as file:        ## rb -- read the binary file
     content=file.read()

print(content)
```

```python
## read the content from source text file to destination text file

with open("source.txt",mode="r") as source_file:
     content=source_file.read()


with open("destination.txt",mode="w") as destination_file:
     destination_file.write(content)
```

```python
##  read a text file and count the number of lines,words and character

def count_file_content(filepath):
   with open(filepath,mode="r") as file:
    lines = file.readlines()
    lines_count = len(lines)
    words_count = sum(len(line.split()) for line in lines)
    char_count = sum(len(line) for line in lines)
   return lines_count,words_count,char_count

filepath="example.txt"
lines_count,words_count,char_count=count_file_content(filepath)
print(f'no_of_lines:{lines_count},no_of_words:{words_count},char_count:{char_count}')

```

Working with file path
----------------------------

```python
## create a new directory
import os

new_directory="package"
os.mkdir(new_directory)
print(f"{new_directory} created")
```

```python
## listing files and directory

item=os.listdir('.')     # . represent the current directory
print(item)
```

```python
## joining the paths -- important in real time project

dir_name="folder"
file_name="file.txt"
full_path=os.path.join(dir_name,file_name)
print(full_path)
```

```python
## if we want to get current working directory

full_path_cur=os.path.join(os.getcwd(),dir_name,file_name)
print(full_path_cur)
```

```python
## checking the file exists

path="example1.txt"
if os.path.exists(path):
  print(f"the path {path} already exists")
else:
  print(f" the path{path} does not exists")
```

```python
# checking if a path is a file or directory

path="example.txt"
if os.path.isfile(path):                      ## isfile
  print(f"{path} is a file")
elif os.path.isdir(path):                     ## isdir
  print(f"{path} is directory")
else:
  print(f"{path} is not a file or directory")
```

```python
## getting the absolute path

realtive_path = "example.txt"

absolute_path = os.path.abspath(realtive_path)
print(absolute_path)
```

Expections handling
-------------------
Expection handling in python allows you to handle errors gracefully and take corrective actions without stopping the execution of the program.

What is expections:
---------------------
Exceptions are events that disrupt the normal flow of program. They occur when an error is encountered during program execution. Common exceptions include,
 --> zeroDivisionError : Dividing by zero
 --> FileNotFoundError : File not found
 --> ValueError: Invalid value



```python
a=b
```

```python
## to handle expections for the above error

try:
  a=b
except:
  print("the variable is not defined")
```

```python
## another way to code above


try:
  a=b
except NameError as ex:
  print(ex)
```

```python
## try,except,else block

try:
  num = int(input("enter a number"))
  result=10/num
except ValueError:
  print("this is not valid number")
except ZeroDivisionError:
  print("enter denominator greater than 0")
except Exception as ex:
  print(ex)
else:
  print(f"the result is  {result}")                   # if the error is not there, then else block will happen
```

```python
## try,except,else and finally

try:
  num = int(input("enter a number"))
  result=10/num
except ValueError:
  print("this is not valid number")
except ZeroDivisionError:
  print("enter denominator greater than 0")
except Exception as ex:
  print(ex)
else:
  print(f"the result is  {result}")
finally:
  print("execution completed")               ## whether the error comes or not , finally block will execute
```

```python
### file handling and exception handling

try:
  file=open("example1.txt",mode="r")
  content = file.read()
  print(content)
except FileNotFoundError:
  print(f"{file} not found")
except Exception as ex:
  print(ex)
finally:
  if 'file' in locals() or not file.closed():
    file.close()
    print('file close')
```
