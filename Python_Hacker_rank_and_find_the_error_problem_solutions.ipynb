HACKER RANK ASSESSMENT

Implement a steming algorithm to extract the base for of words using these rules.
1. If the word ends in "ed","ly" or "ing", reove the suffix
2. If the resulting ord is longer than 8 letters, keep only the first 8 letters

Create a function that takes a string of space seperated word and returns it stemmed counterpart

Example:
text="an extremely dangerous dog is barking"

processing each word:
--> "an" : no suffix to reomve , less than 8 letters -> an
--> "extremely" : Remove "ly" -> "extreme",les than 8 letters ->extreme
--> "dangerous" : no suffix to remove, greater than 8 letters ,
truncate -> dangerou
-->"dog" : no suffix to reomve , less than 8 letters -> dog
-->"is"  : no suffix to reomve , less than 8 letters -> is
-->"barking" : remove "ing" -> "bark",less than 8 letters -> bark


Desired Output:
---------------
"an extreme dangerou dog is bark"

```python
def stemmer(text):
  stemmer_lst=[]
  char_limit=8
  for word in text:
    if word.endswith("ed") or word.endswith("ly") or word.endswith("ing"):
       if word.endswith("ed") or word.endswith("ly"):
          modify_word=word[:-2]
          if len(modify_word) > char_limit:
            modify_word=modify_word[:9]
            stemmer_lst.append(modify_word)
          else:
            stemmer_lst.append(modify_word)
       else:
          modify_word=word[:-3]
          if len(modify_word) > char_limit:
            modify_word=modify_word[:9]
            stemmer_lst.append(modify_word)
          else:
            stemmer_lst.append(modify_word)
    elif len(word) <= char_limit:
          stemmer_lst.append(word)
    else:
        continue
  return " ".join(stemmer_lst)



text=input().split()
stemmer(text)


```

CENTERED DISPLAY GENERATOR
---------------------------
A design package requires a feature to center-align product names dynamically based on a specified display width on a website. Periods are added on both sides of the product name to enure precise aligment within the given width. If the required number of periods is odd, the left side wil have one ore period than the right.

Given the width of the cell and a list of n strings representing product naes,return a list of center-aligned string

EXAMPLE:
--------
width=11
content=["Middle"]

"Middle" has 6 letters and is centered in 11 positions. There should be 5 periods total, 3 on the left and 2 on the right. i.e:- "...Middle.."

Function description:
---------------------
Complete the function center display in te editor with the following parameters.

int width: the required width of every line
str content[n]: the strings to center

Returns:
--------
str[n] : a list of centered string

```python
def center_align_product_name(product_name, display_width):
  center_align_lst=[]
  for prdname in product_name:
      total_padding = display_width - len(prdname)
      if total_padding <= 0:
         center_align_lst.append(prdname)
      else:
         right_padding = total_padding // 2
         left_padding = total_padding - right_padding
         left_periods = '.' * left_padding
         right_periods = '.' * right_padding
         aligned_name = f"{left_periods}{prdname}{right_periods}"
         center_align_lst.append(aligned_name)
  return " ".join(center_align_lst)

product_name=input().split()
display_width=int(input())
center_align_product_name(product_name,display_width)


```

```python
## Progressive time printer with threshold

'''
Program that once executed the programs print hello instantly first, then it prints it after 1 second, then after 2,3 and then
the program prints out the message "End of the loop" and stops
'''
import time

count=0
while True:
  count=count+1
  print("hello")
  time.sleep(count)
  if count > 3:           ## after 3 second it will exit the loop
    print("End of the loop")
    break
```

```python
## File access from url

'''

Write a python code to print out the content from the file which is in url

https://pythonhow.com/media/data/universe.txt

'''

import urllib.request
import pprint

url = "https://pythonhow.com/media/data/universe.txt"  # Replace with your desired URL

try:
    # Open the URL and read the response
    with urllib.request.urlopen(url) as response:
        # Read the response and decode it as UTF-8
        html_content = response.read().decode('utf-8')
        print("Below is the content from the file")
        pprint.pprint(html_content)

except urllib.error.URLError as e:
    # Handle URL errors (e.g., website not found)
    e.reason="website not found"
    print(f"An error occurred: {e.reason}")
```

```python
# letter count from the file in the url

'''
write a python code to count the number of 'a' characters in the text file which is in url

https://pythonhow.com/media/data/universe.txt

'''

import urllib.request
import pprint

url='https://pythonhow.com/media/data/universe.txt'

try:
  with urllib.request.urlopen(url) as response:
    count=0
    content=response.read().decode('utf-8')
    for char in content:
       if char == 'a':
          count=count+1
       else:
          continue
    print(count)
except urllib.error.URLError as e:
  e.reason="the ebsite is not found"
  print(f"An error occured , reason is {e.reason}")
```

```python
# password generator

'''
create a python script that generates a password of 6 random alphanumeric characters in the range
acdefghijlmnopqrstuvwxyz0123456789ABCDEFGHIJLMNOPQRSTUVXYZ!@#$%^&*()?

'''

import random

def password_generator():
    characters = "acdefghijlmnopqrstuvwxyz0123456789ABCDEFGHIJLMNOPQRSTUVXYZ!@#$%^&*()?"
    chosen = random.sample(characters,12)
    password="".join(chosen)
    return password

while True:
    user = input("pls give the random password yes/no?")
    if user == "yes":
      print(password_generator())
    else:
      print("the user said no , so stoping to give random password")
      break

```

```python
## CREATE A PRG THAT ASKS THE USER TO SUBMIT TEXT THROUGH A GUI

from tkinter import *

window = TK()

file = open("user_gui.txt","a+")

def add():
  file.write(user_value.get() + "\n")
  entry.delete(0,END)

def save():
  global file
  file.close()
  file = open("user_gui.txt","a+")

def close():
  file.close()
  window.destroy()


user_value = StringVar()
entry = Entry(Window, textvariable=user_value)
entry.grid(row=0,column=0)

button_add = Button(window,text="Add line",command=add)
button_add.grid(row=0,column=1)

button_save = Button(window, text="save changes", command=save)
button_save.grid(row=0,column=2)

button_close =Button(window,text="save and close",command=close)
button_close.grid(row=0,column=3)

window.mainloop()
```
